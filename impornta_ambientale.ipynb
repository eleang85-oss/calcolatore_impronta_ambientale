```python
"""
Calcolatore Impronta Ambientale e Spreco Personale
Basato sui parametri di consumo medio e impatto in Italia.
"""

# --- COEFFICIENTI E BENCHMARK UFFICIALI ---
BENCHMARK = {
    "acqua_giorno_media" : 200,        # Litri/giorno (media Italia 174-220 l)
    "luce_anno_media": 1200,           # KWh/anno (media single)
    "gas_anno_medio": 350,             # Sm3/anno (media single)
    "cibo_sprecato_sett_medio": 555.8, # g/Settimana (~29 kg/anno pro capite)

    # Fattori di conversione ambientali
    "co2_per_kwh": 0.28,               # kg CO2e per kWh elettrico
    "co2_per_sm3": 2.0,                # kg CO2e per Sm3 gas metano
    "co2_per_kg_cibo": 2.5,            # kg CO2e per kg cibo sprecato
    "acqua_per_kt_cibo": 1000,         # Litri di acqua virtuale sprecata per cibo
    "biocapacita_sostenibile_gha": 1.6, # Ettari globali sostenibili per persona
    "impronta_media_italia_gha": 3.8   # Ettari globali medi in Italia (~2.3 pianeti)
}

def calcola_impatto(acqua_giorno, luce_kwh, gas_sm3, cibo_g_sett):
  """Calcola le emissioni di CO2 e lo spreco idrico annuale."""
  cibo_kg_anno = (cibo_g_sett * 52) / 1000.0

  co2_luce = luce_kwh * BENCHMARK["co2_per_kwh"]
  co2_gas = gas_sm3 * BENCHMARK["co2_per_sm3"]
  co2_cibo = cibo_kg_anno * BENCHMARK["co2_per_kg_cibo"]
  co2_totale = co2_luce + co2_gas + co2_cibo

  acqua_diretta_anno = acqua_giorno * 365
  acqua_sprecata_cibo = cibo_kg_anno * BENCHMARK["acqua_per_kt_cibo"]

  # Stima proporzionale dell'impronta ecologica in ettari globali (gha)
  rapporto_consumo = (co2_totale / 1500)  # ~1500 kg di CO2e rappresenta un profilo medio
  gha_stimati = round(BENCHMARK["impronta_media_italia_gha"] * rapporto_consumo, 1)
  pianeti_equivalenti = round(gha_stimati / BENCHMARK["biocapacita_sostenibile_gha"], 1)

  return {
      "cibo_kg_anno": round(cibo_kg_anno, 1),
      "co2_totale_kg": round(co2_totale, 1),
      "co2_luce": round(co2_luce, 1),
      "co2_gas": round(co2_gas, 1),
      "co2_cibo": round(co2_cibo, 1),
      "acqua_diretta_anno": round(acqua_diretta_anno, 1),
      "acqua_sprecata_cibo": round(acqua_sprecata_cibo, 1),
      "gha_stimati": gha_stimati,
      "pianeti_equivalenti": pianeti_equivalenti
  }

def simulazione_what_if(cibo_kg_anno):
  """Simula il risparmio ambientale riducendo lo spreco alimentare del 50%."""
  risparmio_cibo_kg = cibo_kg_anno / 2.0
  co2_evitata = risparmio_cibo_kg * BENCHMARK["co2_per_kg_cibo"]
  acqua_salvata = risparmio_cibo_kg * BENCHMARK["acqua_per_kt_cibo"]

  return round(co2_evitata, 1), round(acqua_salvata)

def main():
  print("=" * 60)
  print("🌱 CALCOLATORE IMPATTO AMBIENTALE E SPRECO PERSONALE")
  print("=" * 60)
  print("Inserisci i tuoi dati di consumo (premi Invio per i valori medi):\n")

  try:
    inp_acqua = input(f"Acqua diretta (Litri/giorno) [media : {BENCHMARK ['acqua_giorno_media']}]: ")
    acqua = float(inp_acqua) if inp_acqua.strip() else BENCHMARK["acqua_giorno_media"]

    inp_luce = input(f"Energia elettrica (KWh/anno) [media single]: {BENCHMARK['luce_anno_media']}")
    luce = float(inp_luce) if inp_luce.strip() else BENCHMARK["luce_anno_media"]

    inp_gas = input(f"Gas metano (Sm3/anno) [media single]: {BENCHMARK['gas_anno_medio']}")
    gas = float(inp_gas) if inp_gas.strip() else BENCHMARK["gas_anno_medio"]

    inp_cibo = input(f"Cibo sprecato (grammi/settimana) [media: {BENCHMARK['cibo_sprecato_sett_medio']}]: ")
    cibo = float(inp_cibo) if inp_cibo.strip() else BENCHMARK["cibo_sprecato_sett_medio"]

  except ValueError:
    print("\n❌ Errore: Inserisci numeri validi.")
    return

  # Esecuzione calcoli
  res = calcola_impatto(acqua, luce, gas, cibo)
  co2_evitata, acqua_salvata = simulazione_what_if(res["cibo_kg_anno"])

  # Stampa Report
  print("\n" + "=" * 60)
  print("📊 RISULTATI DEL TUO IMPATTO AMBIENTALE ANNUALE")
  print("=" * 60)
  print(f"• Cibo sprecato:                  {res['cibo_kg_anno']} kg/anno")
  print(f"• Emmissioni di CO₂e Totali:      {res['co2_totale_kg']} kg CO₂e/anno")
  print(f"  ├─ Da consumo elettico:         {res['co2_luce']} kg CO₂e")
  print(f"  ├─ Da consumo di gas:           {res['co2_gas']} kg CO₂e")
  print(f"  └─ Da di cibo sprecato:         {res['co2_cibo']} kg CO₂e")
  print(f"• Acqua diretta consumata:        {res['acqua_diretta_anno']:,} Litri/anno")
  print(f"• Acqua sprecata nel cibo:        {res['acqua_sprecata_cibo']:,} Litri/anno (acqua virtuale)")
  print(f"• Impronta Ecologica:            ~{res['gha_stimati']} gha ({res['pianeti_equivalenti']} pianeti Terra)")

  print("\n" + "=" * 60)
  print("💡 SCENARIO 'WHAT-IF': Dimezza il tuo spreco alimentare")
  print("=" * 60)
  print(f"Se riduci del 50% il cibo che getti via:")
  print(f" --> Eviti {co2_evitata} kg di CO₂e all'anno")
  print(f" --> Risparmi {acqua_salvata:,} litri di acqua virtuale all'anno")
  print("=" * 60)

if __name__ == "__main__":
  main()

```
